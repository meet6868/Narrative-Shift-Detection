# 📊 CSV Data Explorer
## Load, Explore, and Analyze CSV Files

**Features:**
- ✅ Load any CSV file
- ✅ Display column names and data types
- ✅ Show basic statistics
- ✅ Display sample data (first/last rows)
- ✅ Check for missing values
- ✅ Display dataset shape and memory usage

---

## Step 1: Import Required Libraries

```python
# Import libraries
import pandas as pd
import numpy as np
from pathlib import Path
import warnings

warnings.filterwarnings('ignore')

print("✅ Libraries imported successfully!")
print(f"   Pandas version: {pd.__version__}")
print(f"   NumPy version: {np.__version__}")
```

## Step 2: Set File Path

**Instructions:**
- Update the `CSV_FILE_PATH` variable with your CSV file path
- Can be absolute path: `/home/user/data/file.csv`
- Or relative path: `../DATA/file.csv`

```python
# ============================================================================
# SET YOUR CSV FILE PATH HERE
# ============================================================================

# Example paths (uncomment and modify one):
# CSV_FILE_PATH = '../DATA/CNN_Articels_clean.csv'
# CSV_FILE_PATH = '../Processed_Data/ALL_Combined_Data.csv'
# CSV_FILE_PATH = '/home/hp/SEM2/INLP/Naretve_Shift/DATA/Covid_News.csv'

# CSV_FILE_PATH = '../DATA/CNN_Articels_clean.csv'  # UPDATE THIS!

# CSV_FILE_PATH = '../DATA/CNN_Articels_clean.csv'  # UPDATE THIS!
CSV_FILE_PATH = '../Processed_Data/Topic_Wise_w3/Health.csv'  # UPDATE THIS!
# Convert to Path object
csv_path = Path(CSV_FILE_PATH)

# Verify file exists
if csv_path.exists():
    file_size_mb = csv_path.stat().st_size / (1024 * 1024)
    print(f"✅ File found: {csv_path.name}")
    print(f"   Path: {csv_path.absolute()}")
    print(f"   Size: {file_size_mb:.2f} MB")
else:
    print(f"❌ ERROR: File not found at {csv_path.absolute()}")
    print(f"\n💡 Please update the CSV_FILE_PATH variable with the correct path")
```

## Step 3: Load CSV File

```python
# Load CSV file
print("🔄 Loading CSV file...")

try:
    df = pd.read_csv(csv_path)
    print(f"✅ CSV file loaded successfully!")
    print(f"\n📊 Dataset Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
except Exception as e:
    print(f"❌ Error loading file: {e}")
    df = None
```

## Step 4: Display Column Information

```python
if df is not None:
    print("=" * 80)
    print("📋 COLUMN INFORMATION")
    print("=" * 80)
    
    print(f"\n🔢 Total Columns: {len(df.columns)}\n")
    
    # Create column info dataframe
    col_info = pd.DataFrame({
        'Column Name': df.columns,
        'Data Type': df.dtypes.values,
        'Non-Null Count': df.count().values,
        'Null Count': df.isnull().sum().values,
        'Null %': (df.isnull().sum() / len(df) * 100).round(2).values
    })
    
    print(col_info.to_string(index=False))
    print("\n" + "=" * 80)
```

## Step 5: Display Sample Data - First Rows

```python
if df is not None:
    print("=" * 80)
    print("📄 FIRST 5 ROWS (HEAD)")
    print("=" * 80)
    print()
    
    # Set pandas display options for full width
    pd.set_option('display.max_colwidth', None)  # Show full column width
    pd.set_option('display.width', None)  # Auto-detect display width

    display(df.head(1)[['date','War','Health','Climate','Economics','Technology','main_sentence','w3_embedding']])
```

## Step 6: Display Sample Data - Last Rows

```python
if df is not None:
    print("=" * 80)
    print("📄 LAST 5 ROWS (TAIL)")
    print("=" * 80)
    print()
    display(df.tail())
```

## Step 7: Display Random Sample

```python
if df is not None:
    print("=" * 80)
    print("🎲 RANDOM SAMPLE (5 rows)")
    print("=" * 80)
    print()
    display(df.sample(min(5, len(df))))
```

## Step 8: Basic Statistics for Numerical Columns

```python
if df is not None:
    print("=" * 80)
    print("📈 NUMERICAL STATISTICS")
    print("=" * 80)
    print()
    
    numerical_cols = df.select_dtypes(include=[np.number]).columns
    
    if len(numerical_cols) > 0:
        display(df[numerical_cols].describe())
    else:
        print("ℹ️  No numerical columns found")
```

## Step 9: Statistics for Text/Object Columns

```python
if df is not None:
    print("=" * 80)
    print("📝 TEXT/OBJECT COLUMN STATISTICS")
    print("=" * 80)
    print()
    
    object_cols = df.select_dtypes(include=['object']).columns
    
    if len(object_cols) > 0:
        for col in object_cols:
            print(f"\n🔹 Column: {col}")
            print(f"   Unique values: {df[col].nunique():,}")
            print(f"   Most common: {df[col].mode().iloc[0] if len(df[col].mode()) > 0 else 'N/A'}")
            print(f"   Sample values: {list(df[col].dropna().unique()[:3])}")
    else:
        print("ℹ️  No text/object columns found")
```

## Step 10: Missing Values Analysis

```python
if df is not None:
    print("=" * 80)
    print("🔍 MISSING VALUES ANALYSIS")
    print("=" * 80)
    print()
    
    missing_data = pd.DataFrame({
        'Column': df.columns,
        'Missing Count': df.isnull().sum().values,
        'Missing %': (df.isnull().sum() / len(df) * 100).round(2).values,
        'Data Type': df.dtypes.values
    })
    
    # Sort by missing percentage (descending)
    missing_data = missing_data.sort_values('Missing %', ascending=False)
    
    print(missing_data.to_string(index=False))
    
    total_missing = df.isnull().sum().sum()
    total_cells = df.shape[0] * df.shape[1]
    missing_pct = (total_missing / total_cells * 100)
    
    print(f"\n📊 Overall Missing Data:")
    print(f"   Total missing cells: {total_missing:,} out of {total_cells:,} ({missing_pct:.2f}%)")
```

## Step 11: Memory Usage

```python
if df is not None:
    print("=" * 80)
    print("💾 MEMORY USAGE")
    print("=" * 80)
    print()
    
    memory_usage = df.memory_usage(deep=True)
    total_memory_mb = memory_usage.sum() / (1024 * 1024)
    
    print(f"Total memory usage: {total_memory_mb:.2f} MB\n")
    
    # Top memory consuming columns
    memory_by_col = memory_usage.sort_values(ascending=False)
    print("Top 10 columns by memory usage:")
    for col, mem in memory_by_col.head(10).items():
        if col != 'Index':
            print(f"   {col}: {mem / (1024 * 1024):.2f} MB")
```

## Step 12: Complete Dataset Info Summary

```python
if df is not None:
    print("=" * 80)
    print("📊 COMPLETE DATASET INFORMATION")
    print("=" * 80)
    print()
    
    df.info()
```

## Step 13: Export Summary Report (Optional)

```python
if df is not None:
    # Create summary report
    report = {
        'File Name': csv_path.name,
        'File Path': str(csv_path.absolute()),
        'Total Rows': df.shape[0],
        'Total Columns': df.shape[1],
        'Memory Usage (MB)': f"{df.memory_usage(deep=True).sum() / (1024 * 1024):.2f}",
        'Total Missing Values': df.isnull().sum().sum(),
        'Missing Percentage': f"{(df.isnull().sum().sum() / (df.shape[0] * df.shape[1]) * 100):.2f}%",
        'Numerical Columns': len(df.select_dtypes(include=[np.number]).columns),
        'Text Columns': len(df.select_dtypes(include=['object']).columns),
        'Datetime Columns': len(df.select_dtypes(include=['datetime']).columns)
    }
    
    print("=" * 80)
    print("📋 SUMMARY REPORT")
    print("=" * 80)
    print()
    
    for key, value in report.items():
        print(f"{key:.<40} {value}")
    
    print("\n" + "=" * 80)
    print("✅ CSV exploration complete!")
    print("=" * 80)
```
